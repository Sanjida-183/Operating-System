```python
processes_ID=["P1","P2","P3","P4","P5"]
arrival_time=[3,2,5,0,1]
burst_time=[3,1,2,3,2]

completion_time=[0,0,0,0,0]
turnaround_time=[0,0,0,0,0]
waiting_time=[0,0,0,0,0]

order=[3,4,1,0,2]

time=0

for i in order:
    time=time+burst_time[i]
    completion_time[i]=time
    turnaround_time[i]=completion_time[i]-arrival_time[i]
    waiting_time[i]=turnaround_time[i]-burst_time[i]
display("Process AT BT CT TAT WT")

for i in range(len(processes_ID)):
    display(processes_ID[i],arrival_time[i],burst_time[i],completion_time[i],turnaround_time[i],waiting_time[i])
```
